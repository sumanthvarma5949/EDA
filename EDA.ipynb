# **LINE CHART → MOMTHLY REVENUE**

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("cleaned_sales.csv")
df["month"]=pd.to_datetime(df["Order_Date"]).dt.to_period("M")
a=df.groupby("month")["Total_Sales"].sum()
print(a)
print("-----------------------------------------------------DATA_VISUALIZATION------------------------------------------------------")
months=a.index.astype(str)
revenue=a.values
plt.plot(months,revenue,marker="*")
plt.xlabel("months")
plt.ylabel("sales in lakhs ")
plt.title("Line chart → Monthly revenue")
plt.xticks(rotation=45,color="r")
plt.grid(True,linestyle="--")
plt.show()
```

# **LINE CHART→ YEARLY REVENUE**

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("cleaned_sales.csv")
df["YEAR"]=pd.to_datetime(df["Order_Date"]).dt.to_period("Y")
a=df.groupby("YEAR")["Total_Sales"].sum()
print(a)
print("-----------------------------------------------------DATA_VISUALIZATION------------------------------------------------------")
YEARS=a.index.astype(str)
revenue=a.values
plt.plot(YEARS,revenue,marker="*")
plt.xlabel("YEARS")
plt.ylabel("sales in lakhs ")
plt.title("Line chart → YEARLY revenue")
plt.xticks(rotation=45,color="r")
plt.grid(True,linestyle="--")
plt.show()
```

# **BAR CHART → TOP 10 PRODUCTS**

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("cleaned_sales.csv")
A=df.groupby("Product")["Total_Sales"].sum().sort_values(ascending=False)
print(A)

plt.bar(A.index,A.values)
plt.xticks(rotation=45)
plt.show()
```

# **HBAR CHART → TOP 10 PRODUCTS**

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("cleaned_sales.csv")
A=df.groupby("Product")["Total_Sales"].sum().sort_values(ascending=False)
print(A)

plt.barh(A.index,A.values)
plt.xticks(rotation=45)
plt.show()
```

# **PIE CHRT → REGION-WISE SALES**

```python
import pandas as pd
df=pd.read_csv("cleaned_sales.csv")
a=df.groupby("Region")["Total_Sales"].sum()
print(a)
import matplotlib.pyplot as plt
plt.pie(a.values, labels=a.index,autopct="%1.1f%%")#,startangle=90,labeldistance=1 )
# plt.legend()
# plt.axis("equal")
plt.show()
```

# **HISTOGRAM → ORDER QUANTITY DISTRIBUTION**

```python


import pandas as pd
df=pd.read_csv("cleaned_sales.csv")
a=df.groupby("Quantity")["Quantity"].count ()
print(a)
import matplotlib.pyplot as plt
plt.hist(df["Quantity"],bins=25,edgecolor="red")
plt.show()
```

# **BOXPLOT → PROFIT MARGIN DISTRIBUTION**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("cleaned_sales.csv")

# Calculate Profit Margin (%)
df['Profit_Margin'] = df['Profit'] / df['Total_Sales'] * 100

# Create boxplot
plt.figure(figsize=(6,4))
plt.boxplot(df['Profit_Margin'].dropna())
plt.title('Profit Margin Distribution')
plt.ylabel('Profit Margin (%)')
plt.show()
```
